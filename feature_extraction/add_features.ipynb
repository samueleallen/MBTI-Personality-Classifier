# Feature Generator
Here, we extract important features from our text dataset.

```python
import sys

parent_dir_path = '..'
if parent_dir_path not in sys.path:
    sys.path.insert(0, parent_dir_path)

from mypytable import MyPyTable
import text_utils
```

```python
# Load initial, uncleaned csv
mbti = MyPyTable()

mbti.load_from_file("mbti_1.csv")

# Now split up 'posts' column for data cleaning purposes
column_data = mbti.get_column("posts")

new_column_data = text_utils.split_tweets(column_data) 

mbti.add_col("Split Posts", new_column_data)
```

```python
# Extract new features from 'Split Posts' for classification purposes
mbti.add_col("Word Count", text_utils.get_word_count(mbti.get_column("Split Posts")))

mbti.add_col("Polarity", text_utils.get_polarity(mbti.get_column("Split Posts")))

mbti.add_col("TTR", text_utils.get_ttr(mbti.get_column("Split Posts")))

mbti.add_col("Subjectivity", text_utils.get_subjectivity(mbti.get_column("Split Posts")))
```

```python
# Get tag ratio columns
vr, ar, pr, tr = text_utils.get_tag_ratios(mbti.get_column("Split Posts"), mbti.get_column("Word Count"))
mbti.add_col("Verb Ratio", vr)
mbti.add_col("Adjective Ratio", ar)
mbti.add_col("Personal Pronoun Ratio", pr)
mbti.add_col("Third-Person Pronoun Ratio", tr)
```

```python
# Drop split posts column for file saving since it looks ugly
mbti.drop_column("Split Posts")

mbti.save_to_file("cleaned.csv")
```
